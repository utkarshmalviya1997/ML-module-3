Q1. What is Simple Linear Regression?

ans.  Simple Linear Regression is a statistical method that allows us to understand the relationship between two continuous variables: a dependent variable and an independent variable. It models this relationship using a straight line, aiming to predict the value of the dependent variable based on the independent variable.

Q2. What are the key assumptions of Simple Linear Regression?

ans-  The key assumptions of Simple Linear Regression are:

1.  Linearity: The relationship between the independent and dependent variables is linear.
2.  Independence of Errors: The errors (residuals) are independent of each other.
3.  Homoscedasticity: The variance of the errors is constant across all levels of the independent variable.
4.  Normality of Errors: The errors are normally distributed.

Q3. What is heteroscedasticity, and why is it important to address in regression
models?

ans-  Heteroscedasticity refers to the situation where the variability of the errors (residuals) in a regression model is not constant across all levels of the independent variables. This violates one of the key assumptions of ordinary least squares regression.

Q4.  What is Multiple Linear Regression?

ans-  Multiple Linear Regression is an extension of simple linear regression that models the relationship between a dependent variable and two or more independent variables. It aims to predict the value of the dependent variable based on the combined influence of multiple predictors.

Q5. : What is polynomial regression, and how does it differ from linear
regression?

ans-  Polynomial regression is a form of regression analysis in which the relationship between the independent variable and the dependent variable is modeled as an nth-degree polynomial. It differs from linear regression by allowing for non-linear relationships between variables.

Q6. Implement a Python program to fit a Simple Linear Regression model to
the following sample data:
● X = [1, 2, 3, 4, 5]
● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
Plot the regression line over the data points.

ans-  
###
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Reshape X to a 2D array
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])


model = LinearRegression()

model.fit(X, Y)


Y_pred = model.predict(X)


plt.figure(figsize=(8, 6))
plt.scatter(X, Y, color='blue', label='Actual Data Points')
plt.plot(X, Y_pred, color='red', label='Regression Line')
plt.title('Simple Linear Regression')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()


print(f"Intercept: {model.intercept_:.2f}")
print(f"Coefficient: {model.coef_[0]:.2f}")

Q7.Fit a Multiple Linear Regression model on this sample data:
● Area = [1200, 1500, 1800, 2000]
● Rooms = [2, 3, 3, 4]
● Price = [250000, 300000, 320000, 370000]
Check for multicollinearity using VIF and report the results.

ans-  
###
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from statsmodels.stats.outliers_influence import variance_inflation_factor


Area = np.array([1200, 1500, 1800, 2000])
Rooms = np.array([2, 3, 3, 4])
Price = np.array([250000, 300000, 320000, 370000])


data = pd.DataFrame({
    'Area': Area,
    'Rooms': Rooms
})


model_multi = LinearRegression()
model_multi.fit(data, Price)

print("Multiple Linear Regression Model Coefficients:")
print(f"Intercept: {model_multi.intercept_:.2f}")
for i, col in enumerate(data.columns):
    print(f"Coefficient for {col}: {model_multi.coef_[i]:.2f}")
print("\n")


print("Variance Inflation Factor (VIF) results:")
vif_data = pd.DataFrame()
vif_data['feature'] = data.columns
vif_data['VIF'] = [variance_inflation_factor(data.values, i) for i in range(data.shape[1])]
print(vif_data)

print("\nInterpretation of VIF:")
print("- VIF = 1: No correlation between the predictor and other predictors.")
print("- VIF between 1 and 5: Moderately correlated.")
print("- VIF > 5 or > 10: Highly correlated, indicating potential multicollinearity problems.")

Q8. Implement polynomial regression on the following data:
● X = [1, 2, 3, 4, 5]
3
● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
Fit a 2nd-degree polynomial and plot the resulting curve.

ans-  
###
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression


X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])


poly_features = PolynomialFeatures(degree=2)
X_poly = poly_features.fit_transform(X)


model_poly = LinearRegression()


model_poly.fit(X_poly, Y)


X_plot = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
X_plot_poly = poly_features.transform(X_plot)
Y_plot_pred = model_poly.predict(X_plot_poly)


plt.figure(figsize=(8, 6))
plt.scatter(X, Y, color='blue', label='Actual Data Points')
plt.plot(X_plot, Y_plot_pred, color='red', label='2nd-Degree Polynomial Fit')
plt.title('Polynomial Regression (Degree 2)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()


print("Polynomial Regression Coefficients:")
print(f"Intercept: {model_poly.intercept_:.2f}")

print(f"Coefficients (x^1, x^2, ...): {model_poly.coef_[1:]}")

Q9. Create a residuals plot for a regression model trained on this data:
● X = [10, 20, 30, 40, 50]
● Y = [15, 35, 40, 50, 65]
Assess heteroscedasticity by examining the spread of residuals.

ans-  
###
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
Y = np.array([15, 35, 40, 50, 65])

model_res = LinearRegression()
model_res.fit(X, Y)


Y_pred_res = model_res.predict(X)


residuals = Y - Y_pred_res


plt.figure(figsize=(8, 6))
plt.scatter(Y_pred_res, residuals, color='purple')
plt.axhline(y=0, color='red', linestyle='--') # Add a horizontal line at 0 for reference
plt.title('Residuals Plot')
plt.xlabel('Predicted Y')
plt.ylabel('Residuals')
plt.grid(True)
plt.show()


print("Assessment of Heteroscedasticity:")
print("By examining the residuals plot, we look for a constant spread of residuals across all predicted values of Y.")
print("Ideally, for homoscedasticity, the residuals should be randomly scattered around the horizontal line at zero with a constant variance.")
print("If the spread of residuals changes as predicted Y changes (e.g., forms a cone shape, or a bow-tie shape), it suggests heteroscedasticity.")
print("In this specific plot, with a small number of data points, it's hard to make a definitive conclusion. However, if the plot showed a systematic pattern in the spread of residuals, heteroscedasticity would be indicated.")

Q10.: Imagine you are a data scientist working for a real estate company. You
need to predict house prices using features like area, number of rooms, and location.
However, you detect heteroscedasticity and multicollinearity in your regression
model. Explain the steps you would take to address these issues and ensure a robust
model.

ans-  To address heteroscedasticity, I would primarily use data transformations (e.g., log, square root) on the dependent or independent variables, apply Weighted Least Squares (WLS) if the variance pattern is known, or use robust standard errors to ensure valid inference.

For multicollinearity, I would identify highly correlated predictors using VIF or correlation matrices. Then, I'd consider removing redundant variables, combining them into a new feature, using dimensionality reduction techniques like PCA, or applying regularization methods such as Ridge or Lasso regression. Ultimately, cross-validation and feature engineering are essential for a robust model.
